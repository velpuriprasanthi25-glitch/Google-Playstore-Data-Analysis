```python
import pandas as pd
import numpy as np
import plotly.express as px

# Load data
df = pd.read_csv("E:/DATASETS/Play Store Data (1).csv")

# Understanding dataset
df.head()
```

```python
# Clean installs
df['Installs'] = df['Installs'].str.replace('+', '', regex=False)
df['Installs'] = df['Installs'].str.replace(',', '', regex=False)

df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')

df = df.dropna(subset=['Installs'])
df['Installs'] = df['Installs'].astype(int)
```

```python
# Remove categories starting with A, C, G, S
df = df[~df['Category'].str.startswith(('A', 'C', 'G', 'S'))]

# Get top 5 categories based on total installs
top_categories = (
    df.groupby('Category')['Installs']
    .sum()
    .sort_values(ascending=False)
    .head(5)
    .index
)

# Filter dataset for only top 5 categories
df = df[df['Category'].isin(top_categories)]

# Check result
print(df['Category'].unique())
```
